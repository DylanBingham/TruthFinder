# Functions to retrieve the features of an input article

```python
import pandas as pd
import matplotlib.pyplot as plt
from textblob import TextBlob
import nltk
import numpy as np
import scipy
import sklearn
```

## Get the article text

```python
from goose3 import Goose
```

```python
def get_text(url):
    g = Goose()
    article = g.extract(url=url)
    return article.cleaned_text
```

### Testing

```python
test_article = get_text("https://abcnews.go.com/Technology/wireStory/finding-health-advice-social-media-easier-knowing-claims-120054862")
print(test_article)
```

## Getting rid of any smart quotes

```python
# removing strange characters
def clean_encoding(text):
    text = text = text.replace('â€˜', "'").replace('â€™', "'").replace('â€œ', '"').replace('â€', '"')
    text = text.replace('â€“', '–').replace('â€”', '—').replace('â€', '')
    text = text.replace("“", '"').replace("”", '"').replace("’", "'").replace("‘", "'")
    return text
```

### Testing

```python
test_article = clean_encoding(test_article)
print(test_article)
```

## Get Word Count

```python
def get_article_word_count(article):
    return len(article.split())
```

## Getting Overall Polarity and Subjectivity

```python

def find_polarity(article):
    news = TextBlob(article)
    return news.sentiment.polarity

def find_subjectivity(article):
    news = TextBlob(article)
    return news.sentiment.subjectivity
```

## Getting Average Polarity and Subjectivity by Sentence

```python
def find_avg_polarity_sentences(article):
    news = TextBlob(article)
    polarity = []
    for sentence in news.sentences:
        polarity.append(sentence.sentiment.polarity)
    return sum(polarity) / len(polarity)

def find_avg_subjectivity_sentences(article):
    news = TextBlob(article)
    subjectivity = []
    for sentence in news.sentences:
        subjectivity.append(sentence.sentiment.subjectivity)
    return sum(subjectivity) / len(subjectivity)
```

## Getting Approximate Speech Attributes

##### (number of times "said", or "say" appears in the article)

```python
# testing shows that "say" includes "saying", so no need to count over again.

def get_num_quotes(text):
    text = text.lower()
    return text.count("said") + text.count("say") + text.count("told") + text.count("tell") 
```

```python
def create_article_info(article):
    article_info = {}
    article_info['word_count'] = get_article_word_count(article)
    article_info['polarity'] = find_polarity(article)
    article_info['subjectivity'] = find_subjectivity(article)
    article_info['avg_polarity'] = find_avg_polarity_sentences(article)
    article_info['avg_subjectivity'] = find_avg_subjectivity_sentences(article)
    article_info['num_quotes'] = get_num_quotes(article)
    article_info = pd.DataFrame(article_info, index=[0])
    return article_info
```

```python
def get_article_info(url):
    article = get_text(url)
    article = clean_encoding(article)
    article_info = create_article_info(article)
    return article_info
```
